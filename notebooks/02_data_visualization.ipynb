# 02 — Data Visualization

Show a few ACNE04 images with their annotated lesion bounding boxes.

`lesion_boxes` is stored as parallel lists: `xmin[i], ymin[i], xmax[i], ymax[i]`
describe the i-th box.

```python
import json
from pathlib import Path
import matplotlib.pyplot as plt
import matplotlib.patches as patches
from PIL import Image

PROCESSED = Path('..') / 'data' / 'processed' / 'ACNE04' / 'train'
IMAGES = PROCESSED / 'images'
records = [json.loads(line) for line in (PROCESSED / 'annotations.jsonl').open()]
print('records:', len(records))
```

```python
def show(rec):
    img = Image.open(IMAGES / rec['image_file'])
    fig, ax = plt.subplots(figsize=(6, 6))
    ax.imshow(img)
    boxes = rec.get('lesion_boxes') or {}
    xs, ys = boxes.get('xmin', []), boxes.get('ymin', [])
    xe, ye = boxes.get('xmax', []), boxes.get('ymax', [])
    for x0, y0, x1, y1 in zip(xs, ys, xe, ye):
        ax.add_patch(patches.Rectangle((x0, y0), x1 - x0, y1 - y0,
                                       fill=False, edgecolor='red', linewidth=1.5))
    ax.set_title(f"{rec['image_file']} | severity={rec.get('severity')} | "
                 f"lesions={rec.get('num_lesions_annotated')}")
    ax.axis('off')
    plt.show()

show(records[0])
```
