# 03 — Annotation Analysis

Analyse severity levels, lesion counts and the cross-validation fold splits.

```python
import json
from pathlib import Path
import pandas as pd

PROCESSED = Path('..') / 'data' / 'processed' / 'ACNE04' / 'train'
df = pd.DataFrame(json.loads(l) for l in (PROCESSED / 'annotations.jsonl').open())
df.shape
```

```python
# Severity / level distribution
print(df['severity'].value_counts())
print()
print(df['level'].value_counts().sort_index())
```

```python
# Lesion-count statistics
df['num_lesions_annotated'].describe()
```

```python
# Train/test balance for each of the 5 cross-validation folds
fold_cols = [c for c in df.columns if c.startswith('fold_')]
for c in sorted(fold_cols):
    print(c, dict(df[c].value_counts()))
```
